```python
#Count Vowels and consonants in a string
word="Elephant"
vowels=["a","e","i","o","u","A","E","I","O","U"]
vowel_count=0
consonant_count=0
for i in word:
    if i in vowels:
        vowel_count+=1
    else:
        consonant_count+=1
print("Vowel_count",vowel_count)
print("consonant_count",consonant_count)


```

```python
#Function to find factorial
def fact(n):
    fact=1
    for i in range(1,n+1):
        fact=fact*i
    return fact

n=int(input("enter:"))
print(fact(n))

```

```python
#Count word frequency in a string using a dictionary
str=input("enter a string:")
words=str.split()
frequency={}
for word in words:
    if word in frequency:
        frequency[word]+=1
    else:
        frequency[word]=1
print("Word frequency:",frequency)
```
