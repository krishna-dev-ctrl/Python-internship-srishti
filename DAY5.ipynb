```python
def circle_calculations(radius):
    import math
    area=math.pi*radius**2
    circumference=2*math.pi*radius
    return area,circumference
area,circ=circle_calculations(7)
print(f"Area:{area:.2f}")
print(f"Circumference:{circ:.2f}")
```

```python
def show_args(*args):
    print("args:",args)
def show_kwargs(**kwargs):
    print("kwargs:",kwargs)

show_args(1,2,3)
show_kwargs(name="Bob",age="25")
```

```python
def sum_numbers(*args):
    total=0
    for num in args:
        total+=num
    return total

print(sum_numbers(1,2,3))
print(sum_numbers(10,20,30,40))
print(sum_numbers(5))
```

```python
def create_person(**kwargs):
    person={}
    for key,value in kwargs.items():
        person[key]=value
    return person
person_1=create_person(name="Alice",age=25,city="New York")
person_2=create_person(name="Bob",occupation="Engineer")
print(person_1)
print(person_2)
```

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n*factorial(n-1)
print(factorial(7))
```

```python
def fibonacci(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return fibonacci(n-1)+fibonacci(n-2)
print(fibonacci(6))
```

```python
def recursive_sum(lst):
    if not lst:
        return 0
    else:
        return lst[0]+recursive_sum(lst[1:])
print(recursive_sum([1,2,3,4,5]))
```

```python
counter=0
def increment():
    global counter
    counter+=1
    print("Counter inside function:",counter)
increment()
increment()
print("Counter outside function",counter)
```

```python
def calculate_discount(price:float,discount_rate:float=0.10)->float:
    if price<0:
        raise ValueError("Price cannot be negative")
    if not(0<=discount_rate<=1):
        raise ValueError("Discount rate must be between 0 and 1")
    return price *(1-discount_rate)
final_price=calculate_discount(1200,0.25)
print(f"final price after discount:${final_price:,.2f}")
```
